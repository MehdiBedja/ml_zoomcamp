```python
import numpy as np
#vector to vector
def vectorvecto(u,v) :
    if u.shape[0] != v.shape[0] :
        return
    n= u.shape[0]
    result = 0.0
    for i in range(n) : 
        result = result + u[i]*v[i]
    return result

a = np.ones(5)
b = np.ones(3)
print(vectorvecto(a,b))


c = np.array( [ [1,2,3], [3,2,1] ,[0,1,3]])

c1 = np.linalg.inv(c)
print(c1)
print(c1.dot(c))
print(c.dot(b))
```
